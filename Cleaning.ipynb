```python
# data = pd.read_excel("sales_raw_500.xlsx")
# data
data = pd.read_csv("sales_raw_new_2026.csv")
data
```

```python
print(data.info())
```

```python
print(data.describe())
```

```python
# data= data.isnull()
# data
data = data.drop_duplicates()
data
```

```python
# Standard fix
data['Order_Date'] = pd.to_datetime(data['Order_Date'], errors='coerce')
```

```python
data = data.dropna(subset=['Order_Date'])
```

```python
data
```

```python
# profit
# sales-cost
data['profit']= data['Sales']- data["Cost"]
```

```python
# year column
data['year']= data['Order_Date'].dt.year
```

```python
# Month Column
data['Month']= data['Order_Date'].dt.month
```

```python
print(data)
```

```python
data.to_csv('sales_cleaned.csv', index=False)
```
